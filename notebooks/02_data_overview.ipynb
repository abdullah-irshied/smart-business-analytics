```python
start = perf_counter()
data_first = pd.read_excel(file_path, sheet_name="Year 2009-2010")

print(data_first.shape)
print("Seconds:", round(perf_counter() - start, 2))
```

```python
start = perf_counter()
data_second = pd.read_excel(file_path, sheet_name="Year 2010-2011")

print(data_second.shape)
print("Seconds:", round(perf_counter() - start, 2))
```

```python
data_first.info()
```

```python
data_second.info()
```

```python
print("First sheet:")
print(data_first["InvoiceDate"].min())
print(data_first["InvoiceDate"].max())

print("Second sheet:")
print(data_second["InvoiceDate"].min())
print(data_second["InvoiceDate"].max())
```

```python
overlap_first = data_first[
    data_first["InvoiceDate"] >= data_second["InvoiceDate"].min()
]

print(overlap_first.shape)
```

```python
overlap_second = data_second[
    data_second["InvoiceDate"] <= data_first["InvoiceDate"].max()
]

print(overlap_second.shape)
```

```python
same_overlap = overlap_first.reset_index(drop=True).equals(
    overlap_second.reset_index(drop=True)
)

print(same_overlap)
```

```python
first_check = overlap_first.reset_index(drop=True)
second_check = overlap_second.reset_index(drop=True)

differences = first_check.compare(second_check)
differences.head()
```

```python
overlap_comparison = pd.merge(
    overlap_first.drop_duplicates(),
    overlap_second.drop_duplicates(),
    how="outer",
    indicator=True
)

print(overlap_comparison["_merge"].value_counts())
```

```python
first_counts = overlap_first.value_counts(dropna=False).reset_index(name="first_count")
second_counts = overlap_second.value_counts(dropna=False).reset_index(name="second_count")

count_comparison = pd.merge(first_counts, second_counts)

different_counts = count_comparison[
    count_comparison["first_count"] != count_comparison["second_count"]
]

print(different_counts.shape)
```
